```python
list1 = list(range(5))
list2 = list(range(1, 7))
list3 = list(range(0, 5, 2))
print(list1)
print(list2)
print(list3)
print('최소값', '최대값 :', min(list1), max(list1))
print('합계 :', sum(list1))
```
