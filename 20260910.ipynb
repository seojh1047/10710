```python
# 한 문자만 인덱싱 : 변수명{인덱스번호}
a = 'Hello Python'
print(a[0])
print(a[6])
print(a[-1])
```

```python
a = 'Hello Python'
print(a[0:5])
print(a[6:])
print(a[:5])
print(a[6:-4])
print(a[0:9:2])
```
