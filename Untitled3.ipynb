```python
def get_bmi(h, w):
    return w / ((h / 100) ** 2)

h = float(input("키(cm): "))
w = float(input("몸무게(kg): "))
bmi = get_bmi(h, w)

if bmi < 18.5: status = "저체중"
elif bmi < 23: status = "정상"
elif bmi < 25: status = "과체중"
else: status = "비만"

print(f"BMI: {bmi:.1f} ({status})")
```

```python
def bmi_cal(height, weight):
  bmi_result = weight / (height ** 2)
  return bmi_result

print('키와 몸무게를 공백으로 구분하여 입력하세요(예;165 55)')
h,w = map(float, input( ).split( ))
h = h / 100
bmi = bmi_cal(h, w)
print('당신의 BMI는 %.21입니다.'% bmi)
```

```python
def fac(n):
  if n > 1:
    return n * fac(n - 1)
  else:
    return 1

num = int(input('숫자를 입력하세요: '))
result = fac(num)
print(result)
```
